# Basics

## Arithmetic and Variables
Variables bind names to values; arithmetic operators manipulate numeric values.

```python
# Variable assignment
x = 10         # int
y = 5.5        # float
name = "AI"    # string
is_ready = True  # boolean

(x, y, name, is_ready)
```

```python
# Arithmetic operators
a, b = 12, 5
print('a + b =', a + b)
print('a - b =', a - b)
print('a * b =', a * b)
print('a / b =', a / b)   # true division
print('a // b =', a // b) # floor division
print('a % b =', a % b)   # modulus
print('a ** b =', a ** b) # exponentiation
```

```python
# In-place updates
counter = 0
counter += 1
counter *= 3
counter
```

## Functions
Encapsulate reusable logic, accept parameters, and return values.

```python
# Defining and calling functions
def greet():
    print("Hello, welcome to Python!")

greet()
```

```python
# Parameters and return values
def add_numbers(x, y):
    return x + y

result = add_numbers(3, 7)
result
```

```python
# Default parameters and keyword arguments
def power(base, exponent=2):
    return base ** exponent

print(power(4))       # default square
print(power(4, 3))    # cube
print(power(exponent=4, base=2))
```

```python
# *args, **kwargs and annotations
def summarize(title: str, *items: int, sep: str = ', ', **meta) -> str:
    body = sep.join(map(str, items))
    return f"{title}: {body} | meta={meta}"

summarize('nums', 1, 2, 3, sep=' - ', author='dev', version=1)
```

```python
# Docstrings and help
def normalize(text: str) -> str:
    """Lowercase and strip a string."""
    return text.lower().strip()

help(normalize)
```

## Data Types
Integers, floats, booleans, strings, and type conversion.

```python
# Numeric types and type()
x = 10        # int
y = 3.14      # float
z = 2 + 3j    # complex
(type(x).__name__, type(y).__name__, type(z).__name__)
```

```python
# Booleans and comparisons
is_active = True
print(is_active)
print(5 > 3)
print(5 == 3)
print(bool(0), bool(1), bool(''), bool('text'))
```

```python
# Strings: indexing, slicing, methods
s = ' Python Basics '
print(s[1])
print(s[-2:])
print(len(s))
print(s.strip())
print(s.lower())
print(s.upper())
print(s.replace('Python', 'AI'))
print(s.startswith(' '))
print('::'.join(['a','b','c']))
```

```python
# f-strings and format specifiers
pi = 3.1415926535
name = 'Ada'
print(f'Hello {name}, pi≈{pi:.3f}')
print('{0} in hex is {0:#x}'.format(255))
```

```python
# Type conversion (casting)
num = 5
as_float = float(num)
as_str = str(as_float)
int_from_str = int('42')
(as_float, as_str, int_from_str)
```

## Conditions and Conditional Statements
Control program flow using if/elif/else and boolean logic.

```python
age = 20
if age < 18:
    label = 'Minor'
elif 18 <= age < 60:
    label = 'Adult'
else:
    label = 'Senior'
label
```

```python
# Nested conditions and logical operators
num = 15
if num > 0:
    if num % 2 == 0:
        res = 'Positive Even'
    else:
        res = 'Positive Odd'
else:
    res = 'Non-positive'

check = (0 < num < 20) and (num != 10)
(res, check)
```

```python
# Ternary expression
x = -3
type_label = 'non-negative' if x >= 0 else 'negative'
type_label
```

## Intro to Lists
Ordered, mutable collections with rich slicing and methods.

```python
nums = [1, 2, 3, 4, 5]
mixed = [1, 'AI', 3.14, True]
empty = []
(nums, mixed, empty)
```

```python
# Indexing, slicing
print(nums[0], nums[-1])
print(nums[1:4])
print(nums[:3], nums[3:])
print(nums[::2])
```

```python
# Modification methods
nums[0] = 10
nums.append(6)
nums.insert(1, 20)
nums.extend([7, 8])
nums
```

```python
# Removal and searching
nums.remove(20)
popped = nums.pop()
idx = nums.index(3)
count_1 = nums.count(1)
(nums, popped, idx, count_1)
```

```python
# Sorting and reversing
arr = [3, 1, 4, 1, 5]
arr_sorted = sorted(arr)
arr.sort(reverse=True)
(arr, arr_sorted)
```

```python
# List comprehensions
squares = [x**2 for x in range(6)]
evengreater10 = [x for x in range(30) if (x % 2 == 0 and x > 10)]
[squares, evengreater10]
```

## Dictionaries
Key-value mappings, ideal for structured and labeled data.

```python
user = {'name': 'Ada', 'age': 36}
user['role'] = 'engineer'
print(user['name'])
print(user.get('city', 'N/A'))
user
```

```python
# Common dict operations
keys = list(user.keys())
vals = list(user.values())
items = list(user.items())
('name' in user, keys, vals, items)
```

```python
# Updating and dictionary comprehensions
other = {'age': 37, 'city': 'London'}
user.update(other)
ages = {c: ord(c) for c in 'abc'}
(user, ages)
```

## Tuples
Immutable ordered sequences; useful for fixed records and multiple return values.

```python
point = (3, 4)
x, y = point
coords = (x, y, x**2 + y**2)
(point, coords)
```

```python
# Tuple as dictionary key
locations = { (51.5, -0.1): 'London', (40.7, -74.0): 'New York' }
locations[(51.5, -0.1)]
```

## Loops
Iterate over sequences and ranges; use break/continue wisely.

```python
# For loop with enumerate
animals = ['cat', 'dog', 'fox']
for i, a in enumerate(animals, start=1):
    print(i, a)
```

```python
# While loop with break/continue
n = 0
acc = 0
while n < 10:
    n += 1
    if n % 2 == 0:
        continue
    acc += n
    if acc > 20:
        break
(n, acc)
```

```python
# Loop else clause
for i in range(3):
    print(i)
else:
    print('completed without break')
```

```python
# Iterating dictionaries
for k, v in user.items():
    print(f'{k} -> {v}')
```

## Useful Built-ins
Handy functions for iteration and inspection.

```python
nums = [1, 2, 3, 4]
print(sum(nums))
print(list(map(lambda x: x*2, nums)))
print(list(filter(lambda x: x%2==0, nums)))
print(any(x>3 for x in nums), all(x>0 for x in nums))
```
